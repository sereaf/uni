```python
# Gráfok

#'Első módszer: szótár használata'
d = dict() # vagy d = {}
print(d)
d = {0: [1, 2, 6], 1: [0, 1, 2, 5], 2: [0, 1, 3, 5], 3: [2, 5], 5: [1, 2, 3]} # kulcs: csúcsok; érték: a csúcsba futó élek listája
print(d)
G = DiGraph(d) # egyszerű összetett gráf létrehozása
G.show() # gráf ábrázolása
```

```python
V = [1,2,3,4]
E = [(1,3), (1,1), (1,3)]#, (1,3), (2,1), (2,2), (2,3), (3,1), (3,2), (3,4), (4,3), (4,4)]
G = Graph([V, E],multiedges = False, loops = True)
G.show()
```

```python
#'Harmadik módszer: metódusok'
T = Graph(multiedges = True) # vagy graphs.EmptyGraph()
T.add_edge(0,1)
T.add_vertex(2)
T.add_vertices([7,8])
T.add_edges([(2,1),(7,2),(8,1)])
T.set_edge_label(2,1,'c')
T.show(edge_labels = True)
T.delete_edge((2,1)) # delete_edges([...])
T.delete_vertex(0) # delete_vertices([...])
T.show()
```
