1. 1. Írj egy függvényt, amely az alábbi szabály szerint generál egy sorozatot egy adott a0
   kezdeti értékből, és térjen vissza a periódus hosszával és a periodus elemeivel:
   LaTeX: a\_n\\:=\\frac{\\:a\_{n\-1}}{2}\\:ha\\:a\_{n\-1}\\: páros  és LaTeX: a\_n\\:=\\:\\:3\\cdot a\_{n\-1}\\:\+\\:1 ha LaTeX: a\\:\_{n\-1} páratlan.
   Bemenet: a0
   ​ \(pozitív egész szám\).
   Kimenet: A sorozat periódusa \(értékek listája\) és a periódus hossza.



```sagemath
def periodus(a0):
    sorozat = [a0]
    volt = {}
    while sorozat[-1] not in volt:
        volt[sorozat[-1]] = len(sorozat)
        if sorozat[-1] % 2 == 0:
            sorozat.append(sorozat[-1] / 2)
        else:
            sorozat.append(3 * sorozat[-1] + 1)
    
    _periodus = sorozat[volt[sorozat[-1]]:]
    return _periodus, len(_periodus)

periodus(10)
```

2. 2. Írj egy függvényt, amely kiszámítja az összes Mersenne\-prímet egy adott n határig. Egy Mersenne\-prím olyan prímszám, amely a következő formában írható fel:
   LaTeX: M\_p = 2^p \- 1Mp=2p−1
   , ahol p is prímszám
   Bemenet: n \(pozitív egész szám, a felső határ\).
   Kimenet: Egy lista az összes Mp
   értékből Mp≤n\-re.



```sagemath
def mersenne(n):
    ps = list(primes(n))
    
    mps = []
    for p in ps:
        mp = 2 ^ p - 1
        if mp <= n and is_prime(mp):
            mps.append(mp)
            
    return mps

print(mersenne(10))
```

   \-beli polinomnak. 
   Bemenet: f egészek feletti polinom
   Kimenet:  LaTeX: f\_p \\, \\mathbb{Z} \_pfpZp
   feletti polinom
   b. Írj függvényt ami kiszámolja a legnagyobb közös osztót két polinom esetén Euklideszi algoritmussal LaTeX: \\mathbb{Z} \_pZp
   felett.
   Bemenet: f polinom
   Kimenet: lista



```sagemath
def atalakit(f, p):
    F = GF(p)
    R = PolynomialRing(F, 'x')
    return R(f)

def fgcd(f1, f2, p):
    F = GF(p)
    R = PolynomialRing(F, 'x')
    f1p = R(f1)
    f2p = R(f2)
    return f1p.gcd(f2p)
```
